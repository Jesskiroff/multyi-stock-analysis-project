```python
tickers = ["NVDA", "JNJ", "XOM", "JPM", "WMT", "BTC-USD", "GLD"]
data = yf.download(tickers, start="2022-08-25", end="2025-08-25", threads=False)
data.head()

```

```python
data_clean = data.dropna()
```

```python
data_clean.shape
```

```python
data_clean.isna().sum()
```

```python
close = data_clean["Close"]
close.head()
```

```python
close.shape
```

```python
returns = close.pct_change()
returns.head()
```

```python
returns = close.pct_change().dropna()
returns.head()
```

```python
returns.shape
```

```python
returns.std()
```

```python
annual_vol = returns.std() * np.sqrt(252)
annual_vol
```

```python
annual_return = returns.mean() * 252
annual_return
```

```python
sharpe = annual_return / annual_vol
sharpe.sort_values(ascending=False)
```

```python
returns.corr()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(returns.corr(), annot=True, cmap="coolwarm", center=0)
plt.title("Asset Return Correlations (Aug 2022–Aug 2025)")
plt.show()
```

```python
cumulative = (1 + returns).cumprod()
cumulative.head()
```

```python
cumulative.tail()
```

```python
import matplotlib.pyplot as plt

cumulative.plot(figsize=(12, 6))
plt.title("Growth of $1 Invested (Aug 2022–Aug 2025)")
plt.ylabel("Value of $1")
plt.xlabel("Date")
plt.legend(title="Ticker")
plt.show()
```
